```python
import pandas as pd
space_table = []
newline= '\\newline\\hspace*{.5em}'
space = '{\\hfill}'
s_space = '\\hspace{1em}'
# \pstart[\subsection*{\textcolor{red}{פרק א}}]
perek_pattern = r'\\pstart\[\\subsection\*\{\\textcolor\{red\}\{פרק\s([א-ת]{1,3})\}\}\]'
passuk_pattern = r'\{\\loc\{ ([א-ת]{1,3})\}~\}‏'
current_perek = None
current_passuk = None
perakim = full_text.split(perek_pattern)
for line in lines:
    fill_count = line.count('{\\hfill}')
    line_count = line.count('\\newline\\hspace*{.5em}')
    s_space_count = line.count('\\hspace{1em}')
    perek_match = re.search(perek_pattern, line)
    passuk_match = re.search(passuk_pattern, line)
    if perek_match:
        current_perek = perek_match.group(1)
        current_passuk = None
        # print(perek)
    if passuk_match:
        current_passuk = passuk_match.group(1)
        # print(passuk)
    if current_perek==None or current_passuk==None or line=="" or line=="\\par" or line=="\\ledpb":
        continue
    space_table.append((fill_count, line_count, s_space_count, current_perek+":"+current_passuk))
space_table_df = pd.DataFrame(space_table, columns=['fill_count', 'line_count', 's_space_count',"loc"])
space_table_df['sum']= space_table_df.fill_count + space_table_df.line_count+space_table_df.s_space_count
space_table_df
```

```python
modfied = []
newline= r'\newline\hspace*{.5em}'
space = r'{\hfill}'
s_space = r'\hspace{1em}'
for line in lines:
    fill_count = line.count('{\\hfill}')
    line_count = line.count('\\newline\\hspace*{.5em}')
    splt = line.split('~}')
    loc,txt = None,None
    if len(splt) > 1:
        loc,txt= splt[0],splt[1]
        l_splt= txt.split(newline)
        if fill_count < line_count and loc !="{\loc{ א}":
            print(line.replace(newline, '\n').replace(space, '/\t').replace('\\par', ''))
            # replace last occurance of newline with space
            mod = newline.join(l_splt[:-1]) + space + l_splt[-1]
            mod = loc + '~}' + mod
            modfied.append(mod)
        elif line_count==0 and fill_count==3:
            s_splt = txt.split(space)
            mod = s_splt[0]+space+s_splt[1]+newline+s_splt[2]+space+s_splt[3]
            mod = loc + '~}' + mod
            modfied.append(mod)
        elif l_splt[-1].count(space) != 1 and l_splt[-1].count(s_space) == 1:
            l_splt[-1] = l_splt[-1].replace(s_space, space)
            mod = newline.join(l_splt)
            mod = loc + '~}' + mod
            modfied.append(mod)
        else:
            modfied.append(line)
    else:
        modfied.append(line)
```

```python
import re 
full_text = open('out/MAM-Psalms-tex_noNotes.tex').read().strip()
mod2=[]
perek_pattern = r'\\pstart\[\\subsection\*\{\\textcolor\{red\}\{פרק\s([א-ת]{1,3})\}\}\]'
passuk_pattern = r'\{\\loc\{ ([א-ת]{1,3})\}~\}‏'
newline= r'\newline\hspace*{.5em}'
space = r'{\hfill}'
s_space = "\\hspace{1em}"
spaces = [newline, space, s_space]
all_whitespace = r'(\\newline\\hspace\*\{\.5em\})|(\{\\hfill\})|(\\hspace\{1em\})'
perakim = re.split(perek_pattern,full_text)
mod2.append(perakim[0])
perakim = perakim[1:]
for pi in range(len(perakim))[::2]:
    perek = perakim[pi]
    pesukim = re.split(passuk_pattern,perakim[pi+1])[1:]
    mod2.append( '\\pstart[\\subsection*{\\textcolor{red}{פרק '+perek+'}}]')
    for vi in range(1,len(pesukim))[::2]:
        passuk = pesukim[vi-1]
        v_text = pesukim[vi]
        patterns = re.split(all_whitespace, v_text)
        if patterns:
            text = [s for s in patterns if s]
            patterns = [(p,i) for i,p in enumerate(text) if p in spaces]
            if passuk =="א":
                if len(patterns)>2:
                    for i,p in enumerate(patterns[1:]):
                        if i %2 ==1:
                            text[p[1]] = newline
                        else:
                            text[p[1]] = space
            else:
                if len(patterns)>1:
                    for i,p in enumerate(patterns):
                        if i %2 ==1:
                            text[p[1]] = newline
                        else:
                            text[p[1]] = space
            mod2.append('{\\loc{ '+passuk+'}~}‏'+"".join(text))
        else:
            mod2.append('{\\loc{ '+passuk+'}~}‏'+v_text)
```

```python
mod3=[]
def count_lines(text):
    return text.count(newline) + text.count("\\par")
modfied_text = open('out/MAM-Psalms-tex_noNotes_mod2.tex').read().strip()
perakim = re.split(perek_pattern,modfied_text)
mod3.append(perakim[0])
perakim = perakim[1:]
for pi in range(len(perakim))[::2]:
    mod3.append('\\pstart[\\subsection*{\\textcolor{red}{פרק '+perakim[pi]+'}}]')
    if len(perakim) < pi+2:
        count = count_lines(perakim[pi+1])
        next_count = count_lines(perakim[pi+2])
        if "\\ledpb" in perakim[pi+1] and count + next_count < 25:
            perakim[pi+1] = perakim[pi+1].replace("\\ledpb", "")
        elif "\\ledpb" not in perakim[pi+1] and count + next_count >= 25:
            perakim[pi+1] = perakim[pi+1]+"\\n\\ledpb"
    mod3.append(perakim[pi+1])
with open('out/MAM-Psalms-tex_noNotes_mod3.tex', 'w') as f:
    f.write(''.join(mod3))
```
